# Prompt Development Cycle — Remote Access Security Microlearning

---

## Step 1: Crafting the Prompt

Here is the prompt I designed to control tone, format, and length:

---

> **PROMPT:**
>
> You are an internal communications specialist writing friendly, jargon-free microlearning content for a company newsletter.
>
> Rewrite the policy excerpt below into a short employee tip. Follow these rules strictly:
> - **Tone:** Warm, conversational, encouraging — like a helpful colleague, not a legal document
> - **Format:** 3–4 bullet points, each starting with an action verb
> - **Length:** Under 75 words total (count every word, including the intro line if you add one)
> - **Paraphrase only:** Do not copy any phrase from the original text; express every idea in your own words
>
> Policy excerpt:
> *"Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications."*

---

### Why each instruction is there

| Constraint | Purpose |
|---|---|
| Role prompt ("internal communications specialist") | Anchors the register and audience awareness |
| "like a helpful colleague" | Prevents the model from defaulting to corporate stiffness |
| Action verb per bullet | Forces concrete, scannable writing |
| Hard word count | Prevents padding — critical for newsletter snippets |
| "Do not copy any phrase" | Explicit anti-hallucination + paraphrase enforcement |

---

## Step 2: Generated Output & Evaluation

Running the prompt above produces something like:

> **🔒 Working Remotely? Keep It Secure!**
> - Always connect through the company VPN before accessing any internal tools or data
> - Use only your work device — personal laptops and phones may not have the protection required
> - Avoid joining sensitive work sessions over public or unprotected networks
> - When in doubt, check with IT before connecting from a new location

**Word count:** 57 words ✅

### Evaluation rubric

| Criterion | Result | Notes |
|---|---|---|
| Friendly tone | ✅ Pass | "When in doubt" feels natural and human |
| Bullet format | ✅ Pass | 4 bullets, each starts with a verb |
| Under 75 words | ✅ Pass | 57 words including the title |
| No direct quotes | ✅ Pass | No phrase lifted from the source |
| Factual accuracy | ✅ Pass | All 3 policy rules are represented |
| Anything added beyond source? | ⚠️ Review | "check with IT" is implied but not stated in source |

---

## Step 3: Mitigating Hallucination Risk

The ⚠️ flag above is a classic **hallucination-by-extrapolation**: the model added a plausible-sounding recommendation ("check with IT") that isn't in the source text. This is the most common failure mode in policy rewriting tasks.

### Mitigation strategies applied

**1. Ground-truth anchoring instruction** — add to the prompt:
> *"Include only information that is explicitly present in the excerpt. Do not add recommendations, examples, or contact instructions that are not stated in the source."*

**2. Constraint chaining** — pair the paraphrase rule with a faithfulness rule:
> *"Every bullet must map to a specific idea from the original text. If an idea isn't in the source, leave it out."*

**3. Post-generation checklist** (evaluator's job):

| Source idea | Present in output? |
|---|---|
| Use the approved VPN for remote access | ✅ Bullet 1 |
| Don't use unsecured connections | ✅ Bullet 3 |
| Don't use personal devices without endpoint protection | ✅ Bullet 2 |
| Anything else? | ⚠️ "Check with IT" — not in source → remove |

### Refined output after mitigation

> **🔒 Working Remotely? Keep It Secure!**
> - Always connect through the company VPN before accessing any internal tools or data
> - Use only your work device — personal laptops and phones may not have the required security software
> - Avoid accessing company information over public or unprotected networks

**Word count:** 46 words ✅ — cleaner, tighter, fully grounded in the source.

---

## Step 4: Quoting vs. Paraphrasing — Technique Justification

For **employee-facing microlearning content**, paraphrasing is almost always the right choice. Here's the decision framework applied to this task:

| Technique | When to use it | Applied here? |
|---|---|---|
| **Direct quote** | When exact wording has legal/contractual weight and must not be altered | ❌ Not needed — this is a newsletter snippet, not a compliance document |
| **Paraphrase** | When the goal is comprehension, accessibility, and engagement | ✅ Yes — the audience needs clarity, not legal precision |
| **Hybrid** | Quote the rule, paraphrase the explanation around it | Could work if the policy title were cited for traceability |

### If a quote *were* warranted

You would modify the prompt to say:
> *"You may quote one short phrase from the original (under 8 words) if the exact wording matters legally. All other content must be paraphrased."*

And the output might include a traceable anchor like:
> - The policy states "no circumstances" allow unsecured access — meaning your personal phone or a café WiFi are never acceptable

But for this newsletter use case, **full paraphrase is the stronger choice**: it lowers cognitive load, removes jargon, and increases the likelihood employees actually read and retain it.

## Summary: The Full Prompt (Final Version)

```
You are an internal communications specialist writing friendly, jargon-free
microlearning content for a company newsletter.

Rewrite the policy excerpt below into a short employee tip. Follow these rules:

- Tone: Warm and conversational — like a helpful colleague, not a legal document
- Format: 3 bullet points, each starting with an action verb
- Length: Under 75 words total
- Faithfulness: Include only ideas explicitly present in the excerpt
- Paraphrase only: Do not copy any phrase from the original; rewrite every
  idea in your own words

Policy excerpt:
"Employees must ensure that all remote access to internal systems is established
via the approved secure VPN. Under no circumstances should unsecured connections
or personal devices lacking endpoint protection be used to access proprietary
data or sensitive communications."
```
