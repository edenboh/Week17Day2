## Exercise 1: Open Source Levels Reflection
**Use:** Fully Open, Weights Released, Architecture Only; model components; fine-tuning/domain adaptation; healthcare compliance.

**Step-by-step**
1) Gather definitions for the three openness levels.  
2) Identify key characteristics (what is open; what you can/cannot do).  
3) Compare side by side in a table (what's open vs. impact).  
4) Write a 3?5 sentence comparative paragraph.  
5) Answer the healthcare prompt (1?2 sentences) about retraining on clinical data.

**Deliverables:** paragraph + healthcare answer.

```
# Open Source Levels — Exercise 1

open_source_levels = {
    "Fully Open": {
        "definition": (
            "A model where everything is publicly available: the architecture, "
            "the training code, the pretrained weights, and often the training data. "
            "Examples: LLaMA 2 (partial), Mistral, Falcon."
        ),
        "what_is_open": [
            "Model architecture (layers, attention heads, hyperparameters)",
            "Pretrained weights AND training/fine-tuning code",
        ],
        "what_you_can_do": [
            "Inspect internal mechanisms and reproduce training from scratch",
            "Fine-tune or fully retrain on custom datasets (e.g., clinical data)",
        ],
        "what_you_cannot_do": [
            "Nothing is technically blocked — full control is granted",
            "May still face legal/license restrictions on commercial use",
        ],
    },
    "Weights Released": {
        "definition": (
            "A model where the pretrained weights are made public, allowing "
            "fine-tuning and inference, but the full training pipeline or data "
            "may not be released."
        ),
        "what_is_open": [
            "Pretrained weights (the learned parameters of the model)",
            "Usually the architecture definition (needed to load the weights)",
        ],
        "what_you_can_do": [
            "Run inference directly and fine-tune on domain-specific data",
            "Perform domain adaptation (e.g., medical, legal) via transfer learning",
        ],
        "what_you_cannot_do": [
            "Retrain from scratch — no access to original training code/data",
            "Fully audit or reproduce the pretraining process",
        ],
    },
    "Architecture Only": {
        "definition": (
            "Only the model's structural design is disclosed (number of layers, "
            "attention mechanism type, etc.) without releasing the trained weights "
            "or training code."
        ),
        "what_is_open": [
            "The structural blueprint of the model (layer types, dimensions)",
            "Sometimes the paper describing design choices",
        ],
        "what_you_can_do": [
            "Reimplement the architecture from scratch in your own framework",
            "Train a new model on your own data using the disclosed design",
        ],
        "what_you_cannot_do": [
            "Leverage pretrained knowledge — must train from scratch (costly)",
            "Fine-tune directly — no weights to start from",
        ],
    },
}

for level, info in open_source_levels.items():
    print(f"\n=== {level} ===")
    for key, value in info.items():
        if isinstance(value, list):
            print(f"  {key}:")
            for item in value:
                print(f"    - {item}")
        else:
            print(f"  {key}: {value}")
```

```
comparison_table = """| Openness Level   | What's Open?                                      | Impact on Retraining/Modifying                                                      |
| ---------------- | ------------------------------------------------- | ----------------------------------------------------------------------------------- |
| Fully Open       | Architecture, pretrained weights, training code   | Full control: can inspect, modify, retrain from scratch, or fine-tune               |
| Weights Released | Pretrained weights (+ architecture to load them) | Can fine-tune on new data but cannot reproduce or alter the original training       |
| Architecture Only| Model structure/design only                       | Must train from scratch; no pretrained knowledge to leverage, very resource-costly  |"""
print(comparison_table)
```

```
comparative_paragraph = """
The three open-source levels represent a spectrum of access to a model's internals.
A Fully Open model exposes architecture, weights, and training code alike, giving
practitioners complete freedom to inspect, reproduce, and retrain the model on any
dataset — the gold standard for transparency and adaptability. A Weights Released
model offers pretrained parameters without the full training pipeline, which is
sufficient for fine-tuning and domain adaptation but prevents auditing or reproducing
the original training process. Architecture Only is the most restrictive tier: knowing
the structural blueprint without the weights means starting optimization from scratch,
which demands significant compute and data resources, making it impractical for most
applied teams.
""".strip()
print(comparative_paragraph)
```

```
healthcare_prompt_answer = """
Building a healthcare-specific assistant that must be retrained on clinical data
requires at minimum a Weights Released model, and ideally a Fully Open one.
Weights Released enables domain adaptation through fine-tuning on sensitive clinical
datasets without the prohibitive cost of pretraining from scratch, while the Fully
Open level additionally allows auditing the model's behavior and training pipeline —
a critical requirement for regulatory compliance (e.g., HIPAA) and patient safety.
""".strip()
print(healthcare_prompt_answer)
```
