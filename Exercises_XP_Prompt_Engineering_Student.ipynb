## Exercise 4: Match Prompt to Purpose
Pick a style (Exploratory, Structured, Conversational, Functional) to build a customer-support chatbot prompt.

### Tasks
- Write the full prompt in the chosen style.
- Identify 2 features that prove it matches the style.
- Justify why this style fits better than the others.

```python
chosen_style = 'Conversational'

prompt_template = """
You are Maya, a friendly customer support assistant for ShopEase,
an online retail store.

Your job is to help customers with their orders, returns, and
questions — in a warm, natural, human way.

Guidelines:
- Greet the customer by name if they share it
- Acknowledge their frustration before jumping to solutions
  (e.g., "Oh no, that's really frustrating — let me help you sort this out!")
- Use casual, everyday language — no corporate speak
- Ask one clarifying question at a time if you need more info
- Always end with "Is there anything else I can help you with today? 😊"
- If you can't resolve the issue, say:
  "Let me connect you with a human agent who can take care of this for you."

Current customer message: [INSERT MESSAGE HERE]
"""

key_style_features = [
    "Persona assignment (Maya) — gives the bot a consistent human voice "
    "that customers respond to emotionally",

    "Empathy-first structure — the prompt explicitly instructs the model to "
    "acknowledge feelings before offering solutions, mimicking real human support"
]

justification = """
Conversational style is the best fit here for three reasons:

1. EMOTIONAL CONTEXT — Customer support involves frustrated or confused users.
   A conversational tone builds trust faster than a structured bullet list or
   a purely functional response.

2. FLEXIBILITY — Support queries are unpredictable (returns, delays, wrong items,
   billing...). Conversational prompts adapt naturally to any topic, unlike
   structured templates that feel rigid when the situation doesn't match the format.

3. BRAND VOICE — E-commerce brands need personality. A chatbot that sounds human
   creates a better experience than one that reads like a FAQ page.

Why NOT the others?
- Exploratory → useless here, customers want answers not brainstorming
- Structured  → feels cold and robotic for emotional interactions
- Functional  → great for internal tools (e.g., summarizing a ticket),
                but not for direct customer-facing conversation
"""
```

## Exercise 6: Hallucination Spotting and Mitigation
**Scenario**: Model added an unsupported claim in a climate/marine article summary.

### Tasks
1) Prompt the model to avoid hallucinations using only provided text.
2) Rewrite with verification language.
3) List 2 mitigation strategies.
4) Name 2 domains where hallucinations are risky and why.

```python
# TODO: baseline anti-hallucination prompt
no_hallucination_prompt = """
You are a research assistant summarizing a peer-reviewed scientific article.

Your task: summarize the article provided below in 3-5 bullet points.

STRICT RULES:
- Use ONLY information explicitly written in the article below.
- Do NOT add statistics, claims, or facts from your general knowledge.
- Do NOT infer, extrapolate, or assume anything beyond what is stated.
- If a detail is not in the article, do not include it.
- Copy key figures or percentages exactly as they appear — do not round
  or paraphrase numbers.

[PASTE ARTICLE HERE]
"""
```

```python
# TODO: verification-focused rewrite
verification_prompt = """
You are a fact-grounded research assistant.

Summarize the article below following these verification rules:

1. Only include claims explicitly stated in the article.
   → Before writing each bullet, ask yourself: "Is this sentence
     directly supported by a specific passage in the text?"

2. If a statistic or finding is not clearly stated, write:
   "This information is not available in the provided article."

3. If you are uncertain whether a claim is in the article,
   err on the side of omission — do not guess.

4. After your summary, add a one-line confidence note:
   "All claims above are directly traceable to the source text."
   OR list any points you were uncertain about.

[PASTE ARTICLE HERE]
"""
```
